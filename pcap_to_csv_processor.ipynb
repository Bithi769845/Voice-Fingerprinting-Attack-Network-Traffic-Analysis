# **Load pcapng file and create CSV**

```python
import os
import pandas as pd
from scapy.all import rdpcap

base_path = r'/content/drive/MyDrive/Wireshark'
data = []

print("Starting feature extraction...")

for folder in os.listdir(base_path):
    folder_path = os.path.join(base_path, folder)
    if os.path.isdir(folder_path):
        for file in os.listdir(folder_path):
            if file.endswith('.pcapng'):
                file_path = os.path.join(folder_path, file)
                print(f"Processing: {file}")

                try:
                    packets = rdpcap(file_path)
                    if len(packets) > 0:
                        start_time = packets[0].time
                        end_time = packets[-1].time
                        duration = float(end_time - start_time)

                        lengths = [len(p) for p in packets]
                        avg_len = sum(lengths) / len(lengths)

                        data.append({
                            'Website': folder,
                            'Total_Packets': len(packets),
                            'Duration': duration,
                            'Avg_Packet_Len': avg_len,
                            'Max_Packet_Len': max(lengths)
                        })
                except Exception as e:
                    print(f"Error processing {file}: {e}")

# Create CSV
df = pd.DataFrame(data)
df.to_csv('fingerprinting_dataset.csv', index=False)
print("Dataset created successfully as 'fingerprinting_dataset.csv'!")
```
